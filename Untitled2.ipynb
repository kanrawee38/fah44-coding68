```python
name=input("สวัสดีค่ะชื่ออะไรคะ")
subjects = []
n = int(input("มีกี่วิชาที่ต้องสอบคะ: "))

if n <= 0:
    print("จำนวนวิชาต้องมากกว่า 0")
else:
    for i in range(n):
        subject = input(f"ชื่อวิชาที่ {i+1}: ")
        subjects.append(subject)

    hours_per_day = int(input("เวลาที่คิดว่าจะอ่านค่ะ: "))


    if hours_per_day <= 0:
        print("ชั่วโมงอ่านต้องมากกว่า 0")
    else:
        plan = {}
        hours_each = hours_per_day // n

        if hours_each == 0:
            print("เวลาน้อยเกินไปค่ะ แบ่งไม่พอทุกวิชาค่ะแนะนำให้เพิ่มเวลา")
        else:
            for subject in subjects:
                plan[subject] = hours_each
            print("\n=== ตารางอ่านหนังสือสอบ ===")
            for subject, hour in plan.items():
                print(f"{subject} : อ่าน {hour} ชั่วโมง/วัน")
```
